```python
import numpy as np

def compute_cost(x, y, w, b):
    m = x.shape[0]
    cost = 0.0
    for i in range(m):
        f_wb = w * x[i] + b
        cost += (f_wb - y[i]) ** 2
    total_cost = cost / (2 * m)
    return total_cost

def compute_gradient(x, y, w, b):
    m = x.shape[0]
    dj_dw = 0.0
    dj_db = 0.0
    for i in range(m):
        f_wb = w * x[i] + b
        error = f_wb - y[i]
        dj_dw += error * x[i]
        dj_db += error
    dj_dw = dj_dw / m
    dj_db = dj_db / m
    return dj_dw, dj_db

def gradient_descent(x, y, w1, b1, a1, num_iters):
    w = w1
    b = b1
    J_history = []
    for i in range(num_iters):
        dj_dw, dj_db = compute_gradient(x, y, w, b)
        w = w - a1 * dj_dw
        b = b - a1 * dj_db
        J_history.append(compute_cost(x, y, w, b))

        if i % 100 == 0:
            print(f"Cost = {J_history[-1]:.4f}")

    return w, b, J_history
x_train = np.array([1, 2, 3, 4, 5])
y_train = np.array([2, 4, 6, 8, 10])

w = 0.0
b = 0.0
a = 0.01
iters = 1000

w_final, b_final, J_history = gradient_descent(
    x_train, y_train, w, b, a, iters
)

print(f"w = {w_final:.4f}")
print(f"b = {b_final:.4f}")
```
